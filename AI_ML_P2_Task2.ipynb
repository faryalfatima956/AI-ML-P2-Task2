```python
# Sample Dataset
tickets = [
    "I cannot log into my account.",
    "My payment was deducted twice.",
    "The website is loading very slowly.",
    "I forgot my password and cannot reset it."
]
```

```python
# Zero-Shot Prompt
def zero_shot_prompt(ticket):
  prompt = f"""
  Classify this support ticket and return the top 3 tags.

  Ticket:
  {ticket}
  Return only tags.
  """
  return prompt
```

```python
# Few-Shot Prompt
def few_shot_prompt(ticket):
  prompt = f"""
  Example 1:
  Ticket:I forgot my password.
  Tags:Password Reset, Login Issue, Account Access

  Example 2:
  Ticket:Payment failed during checkout.
  Tags:Payment Issue, Billing, Checkout

  Now Calssify:
  Ticket:{ticket}

  Return top 3 tags only.
  """
  return prompt
```

```python
# Top 3 Tags Output
[
    "Password Reset",
  "Login Issue",
  "Account Access"
]
```

```python
# Comparison
import pandas as pd

comparison = pd.DataFrame({
    "Ticket": [
        "I cannot log into my account",
        "Payment deducted twice",
        "Website loading slowly",
        "Forgot password"
    ],
    "Zero-Shot Tags": [
        "Login, Account, Access",
        "Payment, Billing, Transaction",
        "Performance, Website, Speed",
        "Password, Login, Account"
    ],
    "Few-Shot Tags": [
        "Login Issue, Account Access, Authentication",
        "Payment Issue, Billing Error, Refund",
        "Website Performance, Slow Loading, Technical Issue",
        "Password Reset, Login Issue, Account Access"
    ]
})

comparison
```

The few-shot approach generated more specific and context-aware tags compared to the zero-shot approach. Providing examples helped the model better understand the desired output format and categories.
